```python
import os
import pandas as pd

# IMPORTANT: Replace this with the exact path to the folder shown in your screenshot
data_dir = "path/to/your/dataset/folder"

classes = ["COVID", "Lung_Opacity", "Normal", "Viral Pneumonia"]
data_summary = []

for cls in classes:
    # Target only the 'images' folder inside each class directory
    images_folder_path = os.path.join(data_dir, cls, "images")

    if os.path.exists(images_folder_path):
        # Count all files in the images folder
        num_images = len(os.listdir(images_folder_path))
        data_summary.append({"Class": cls, "Image Count": num_images})
    else:
        print(f"Could not find the 'images' folder at: {images_folder_path}")

# Display the final count
df = pd.DataFrame(data_summary)
print(df)
```
