```python
import os
import pandas as pd
```

```python
data_dir = "/Users/shivam/Documents/Dataset/COVID-19_Radiography_Dataset"

```

```python
classes = ["COVID", "Lung_Opacity", "Normal", "Viral Pneumonia"]
data_summary = []

for cls in classes:
    # Target the 'images' folder inside each class directory
    images_folder_path = os.path.join(data_dir, cls, "images")

    if os.path.exists(images_folder_path):
        # Count all valid image files
        num_images = len([f for f in os.listdir(images_folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
        data_summary.append({"Class": cls, "Image Count": num_images})
    else:
        print(f"Directory not found: {images_folder_path}")

# Generate and print the class distribution table
df = pd.DataFrame(data_summary)
print(df)
```
