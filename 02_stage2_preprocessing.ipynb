```python
import os
import cv2
import albumentations as A

# 1. Setup Input/Output Directories
data_dir = "/Users/shivam/Documents/Dataset/COVID-19_Radiography_Dataset"
output_dir = data_dir + "_Preprocessed"
classes = ["COVID", "Lung_Opacity", "Normal", "Viral Pneumonia"]

# 2. Define Stage 2 Pipeline (Resize + CLAHE only)
preprocess_pipeline = A.Compose([
    A.Resize(width=224, height=224),
    A.CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=1.0)
])

# 3. Process and save all original images
print(f"Saving preprocessed images to: {output_dir}\n")

for cls in classes:
    print(f"Preprocessing class: {cls}...")
    input_class_dir = os.path.join(data_dir, cls, "images")
    output_class_dir = os.path.join(output_dir, cls)
    os.makedirs(output_class_dir, exist_ok=True)

    if os.path.exists(input_class_dir):
        images = [f for f in os.listdir(input_class_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        for img_name in images:
            img_path = os.path.join(input_class_dir, img_name)
            image = cv2.imread(img_path)

            if image is not None:
                # Apply resize and CLAHE
                processed_img = preprocess_pipeline(image=image)['image']
                cv2.imwrite(os.path.join(output_class_dir, img_name), processed_img)

print("\nStage 2 Preprocessing Complete!")
```
